## 5. Problems JW-001 to JW-050

Answers are **not** printed here; they are computed in Section 6. Every problem uses exact rational inputs where possible, SI units internally, and states whether its numbers are taken from a source or are synthetic.

#### JW-001  ·  category A

- **Problem ID:** JW-001
- **Research theme:** Optical communication (carrier frequency)
- **Prerequisites:** none (entry point)
- **Difficulty:** 1
- **Learning objective:** Convert a free-space wavelength to an optical frequency with exact constants.
- **Physical or mathematical context:** Telecom-band light used in time-stretch systems is specified by wavelength; RF engineers think in frequency.
- **Given variables and units:** scalar wavelength λ = 1550 nm = 1.550×10⁻⁶ m; scalar c = 299 792 458 m/s (exact).
- **Unknown variables:** scalar frequency f (Hz).
- **Assumptions:** vacuum wavelength; f > 0, λ > 0.
- **Mathematical model:** c = λ f  (c: m/s, λ: m, f: Hz = 1/s); dimension check: m · s⁻¹ = m · s⁻¹.
- **Question:** Find f exactly as a rational number, then in THz.
- **Expected answer type:** exact rational, then decimal in THz
- **Common mistake:** Using λ in nm without converting to m (off by 10⁹).
- **Verification method:** f·λ reproduces c exactly; Lean `JW.freq_of_wavelength`.
- **Research source or "synthetic educational model":** NIST CODATA / SI 2019 exact constants (c, h, e)

#### JW-002  ·  category A

- **Problem ID:** JW-002
- **Research theme:** Photonics (photon energy)
- **Prerequisites:** JW-001
- **Difficulty:** 1
- **Learning objective:** Compute photon energy in joules and electron-volts.
- **Physical or mathematical context:** Photon energy sets detector bandgap requirements (InGaAs vs Si) for 1550 nm light.
- **Given variables and units:** λ = 1550 nm; h = 6.62607015×10⁻³⁴ J·s (exact); e = 1.602176634×10⁻¹⁹ C (exact).
- **Unknown variables:** scalar E (J) and E (eV).
- **Assumptions:** single photon in vacuum.
- **Mathematical model:** E = h c / λ (J); E_eV = E / e. Dimension: J·s · m/s / m = J.
- **Question:** Compute E in J and eV. Can a silicon detector (bandgap ≈ 1.12 eV) absorb it by band-to-band absorption?
- **Expected answer type:** exact rational + decimal; yes/no
- **Common mistake:** Dividing by e twice or using h-bar instead of h.
- **Verification method:** E_eV · λ(µm) ≈ 1.23984 (the 1240 eV·nm rule).
- **Research source or "synthetic educational model":** NIST CODATA / SI 2019 exact constants (c, h, e)

#### JW-003  ·  category A

- **Problem ID:** JW-003
- **Research theme:** Photonics (photon flux)
- **Prerequisites:** JW-002
- **Difficulty:** 1
- **Learning objective:** Convert optical power to photon arrival rate.
- **Physical or mathematical context:** Shot-noise-limited detection (JW-036) depends on how many photons arrive per pixel.
- **Given variables and units:** P = 1.0 mW = 10⁻³ W; λ = 1550 nm.
- **Unknown variables:** scalar photon rate R (photons/s).
- **Assumptions:** monochromatic light, constant power.
- **Mathematical model:** R = P / E = P λ / (h c). Dimension: W / J = s⁻¹.
- **Question:** Find R. How many photons arrive in a 100 ps time-stretch pixel?
- **Expected answer type:** decimal in scientific notation
- **Common mistake:** Forgetting that E is per photon, so R = P/E, not P·E.
- **Verification method:** R·E = P exactly.
- **Research source or "synthetic educational model":** NIST CODATA / SI 2019 exact constants (c, h, e)

#### JW-005  ·  category J

- **Problem ID:** JW-005
- **Research theme:** Dispersion (β₂ from D)
- **Prerequisites:** JW-001, JW-004
- **Difficulty:** 2
- **Learning objective:** Relate the group-velocity dispersion β₂ to D, with signs.
- **Physical or mathematical context:** Pulse-propagation equations use β₂; datasheets give D.
- **Given variables and units:** D = 17 ps/(nm·km) (typical standard single-mode fiber value at 1550 nm, used here as a textbook number); λ = 1550 nm.
- **Unknown variables:** scalar β₂ in ps²/km.
- **Assumptions:** β₂ and D refer to the same wavelength; D > 0 means anomalous dispersion.
- **Mathematical model:** β₂ = −λ² D / (2π c)  (β₂: s²/m; λ: m; D: s/m²; c: m/s). Dimension: m²·s/m² / (m/s) = s²/m.
- **Question:** Compute β₂ in ps²/km. Is the fiber normal or anomalous at 1550 nm?
- **Expected answer type:** signed decimal
- **Common mistake:** Dropping the minus sign, or mixing ps/(nm km) with SI.
- **Verification method:** D = −2πcβ₂/λ² round trip.
- **Research source or "synthetic educational model":** G. P. Agrawal, *Nonlinear Fiber Optics*, Academic Press, Ch. 3 (dispersion, Gaussian and chirped Gaussian pulses)

#### JW-006  ·  category A

- **Problem ID:** JW-006
- **Research theme:** Time-stretch imaging (frame timing)
- **Prerequisites:** none (entry point)
- **Difficulty:** 1
- **Learning objective:** Use scientific notation to relate frame rate, period and samples per frame.

- **Verification method:** N·(1/f_s) ≤ T_rep < (N+1)/f_s.

#### JW-010  ·  category B

- **Problem ID:** JW-010
- **Research theme:** Phasors (complex optical field)
- **Prerequisites:** none (entry point)
- **Difficulty:** 1
- **Learning objective:** Show that detected intensity does not depend on the absolute optical phase.
- **Physical or mathematical context:** Photodetectors measure |E|²; phase must be recovered indirectly (interference, dispersion).
- **Given variables and units:** scalar complex field E(t) = A e^{i(ωt+φ)} with real A ≥ 0 (V/m), ω (rad/s), φ (rad).
- **Unknown variables:** intensity proxy |E|² (V²/m²).
- **Assumptions:** scalar, monochromatic field.
- **Mathematical model:** |E|² = E · E*.
- **Question:** Simplify |E|² symbolically.
- **Expected answer type:** symbolic expression
- **Common mistake:** Writing |E|² = E² (forgets the conjugate).
- **Verification method:** SymPy simplification (this is algebra, not a formal proof).
- **Research source or "synthetic educational model":** B. E. A. Saleh and M. C. Teich, *Fundamentals of Photonics*, Wiley (interference, Jones matrices, coupled-mode theory, fibers)

#### JW-011  ·  category B

- **Problem ID:** JW-011
- **Research theme:** Interference (fringe visibility)
- **Prerequisites:** JW-010
- **Difficulty:** 2
- **Learning objective:** Derive two-beam interference and fringe visibility.
- **Physical or mathematical context:** Spectral interferometry is a standard way to measure phase in time-stretch systems [GJ13].
- **Given variables and units:** intensities I₁ = 4, I₂ = 1 (arbitrary units); phase difference Δφ (rad).
- **Unknown variables:** I(Δφ), I_max, I_min, visibility V.
- **Assumptions:** mutually coherent, same polarization.
- **Mathematical model:** I = I₁ + I₂ + 2√(I₁I₂) cos Δφ; V = (I_max − I_min)/(I_max + I_min).
- **Question:** Find I_max, I_min and V. Edge case: what is V when I₂ = 0?
- **Expected answer type:** exact numbers
- **Common mistake:** Adding fields' intensities without the cross term.
- **Verification method:** V = 2√(I₁I₂)/(I₁+I₂).
- **Research source or "synthetic educational model":** B. E. A. Saleh and M. C. Teich, *Fundamentals of Photonics*, Wiley (interference, Jones matrices, coupled-mode theory, fibers)

- **Problem ID:** JW-012
- **Research theme:** RF photonics (detector frequency response)
- **Prerequisites:** JW-010
- **Difficulty:** 2
- **Learning objective:** Find magnitude and phase of a first-order transfer function.

## 10. Likely experimental and modeling limitations

- **Second-order dispersion only.** JW-005, -019, -020, -023 and -032 ignore third-order dispersion, loss along the fiber, and nonlinearity. Real time-stretch systems can require all three.
- **Bulk material vs. fiber.** JW-046 uses bulk fused silica; real fiber adds waveguide dispersion, so its group index and D differ.
- **Ideal components.** Lossless polarizers (JW-027), identical couplers (JW-029), single-pole detectors (JW-012, -025), ideal stretch (JW-031) and noiseless unmixing (JW-030).
- **Independence assumptions.** The probability problems (JW-033 to -040) assume independent cells or photons; real flow and real light can be correlated.
- **Synthetic numbers.** Frame rates, data rates, prevalence, sensitivity, cell sizes, and the datasets in JW-028 and JW-045 are synthetic. They teach the method and are not measurements from any laboratory.
- **Approximate material values.** The silicon and silica indices in JW-048 are rounded textbook values; the Cauchy coefficients in JW-014 are synthetic.
